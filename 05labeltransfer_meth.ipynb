```python
import pandas as pd
import plotly.graph_objects as go


df_selected = df_meta_merge[(df_meta_merge.region=='HIP') & (df_meta_merge.NeuronTransmitter=='NN')].copy()

# Group by Subclass and celltype.L2 to count cell transitions
df_counts = df_selected.groupby(['Subclass', 'celltype.L2']).size().reset_index(name='count')

# Create mappings from Subclass and celltype.L2 to numeric indices
all_subclasses = df_selected['Subclass'].unique()
all_celltypes = df_selected['celltype.L2'].unique()

# Assign indices to each category
subclass_to_idx = {subclass: idx for idx, subclass in enumerate(all_subclasses)}
celltype_to_idx = {celltype: idx + len(all_subclasses) for idx, celltype in enumerate(all_celltypes)}

# Generate source (Subclass) and target (celltype.L2) indices
df_counts['source'] = df_counts['Subclass'].map(subclass_to_idx)
df_counts['target'] = df_counts['celltype.L2'].map(celltype_to_idx)

# Extract link parameters
source = df_counts['source'].tolist()
target = df_counts['target'].tolist()
value = df_counts['count'].tolist()

# Combine all labels for nodes
labels = list(all_subclasses) + list(all_celltypes)

fig = go.Figure(data=[go.Sankey(
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=labels,  # Combined labels for Subclass and celltype.L2
        color="blue"   # Customize node colors if needed
    ),
    link=dict(
        source=source,  # Indices of Subclass nodes
        target=target,  # Indices of celltype.L2 nodes
        value=value     # Count of cells flowing between them
    )
)])

fig.update_layout(
    title_text="Cell Flow from Subclass to celltype.L2",
    font_size=10,
    height=1000  # Adjust height for better visibility
)
fig.show()
```
