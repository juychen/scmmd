```python
import scanpy as sc
import numpy as np
```

```python
adata = sc.read_h5ad('/data2st1/junyi/output/atac1112/3REGIONS_peak.h5ads',backed='r')
```

```python
target = 20000
for region in ['PFC','AMY','HIP']:
    obs_region= adata.obs[adata.obs['Region']==region]
    n = obs_region.shape[0]
    if n > target:
        idx = np.random.choice(n, target, replace=False)
        idx_obs = obs_region.index[idx]
    adata_region = adata[idx_obs,:].to_memory()
    adata_region.write_h5ad(f'/data2st1/junyi/output/atac1112/subset/{region}_20k.h5ad')
```
