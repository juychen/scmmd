```python
import pandas as pd
import glob as glob
```

```python
files = glob.glob('/data2st1/junyi/output/atac1112/dar/celltype.L2/MASTNG_dar_annotated_avg/CP_enrichment_results2/*all.csv')
```

```python
df_go = pd.DataFrame()
for f in files:
    df_tmp = pd.read_csv(f)
    df_go = pd.concat([df_go, df_tmp], axis=0)
```

```python
df_go.columns
```

```python
df_go.rename(columns={'Region subclass': 'Region Subclass','Region_subclass_sex_direction': 'Region Subclass Sex Direction','Description':"term_name",'ID':'term_id'}, inplace=True)
```

```python
df_go
```

```python
df_go['source'] = "GO:" + df_go['ONTOLOGY']
df_go['GO'] = df_go['source']+"_"+df_go['term_name']
```

```python
df_go['Region.Subclass'] = df_go['Region Subclass']
df_go['p_value'] = df_go['pvalue']
```

```python
df_go_Up = df_go[df_go['Direction']=='Up']
```

```python
df_go_Up.to_csv('/data2st1/junyi/output/atac1112/dar/celltype.L2/MASTNG_dar_annotated_avg/GO_gsea_summary_all_go_enrichment_Up.csv', index=False)
```

```python
df_go_Down = df_go[df_go['Direction']=='Down']
```

```python
df_go_Down.to_csv('/data2st1/junyi/output/atac1112/dar/celltype.L2/MASTNG_dar_annotated_avg/GO_gsea_summary_all_go_enrichment_Down.csv', index=False)
```
