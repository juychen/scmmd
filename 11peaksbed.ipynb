```python
peaks = pd.read_csv('/data2st1/junyi/output/atac1112/iterative/peaks.csv',index_col=0)
```

```python
peaks['Peaks'].str.split(r'[-:]', expand=True).to_csv('/data2st1/junyi/output/atac1112/iterative/peaks.bed',sep='\t', header=False, index=False)
```

```python
celltypes = peaks.columns.tolist()
celltypes = [ct[:-4] for ct in celltypes][1:]
```

```python
celltypes = list(set(celltypes))
```

```python
outdir = '/data2st1/junyi/output/atac1112/cCRE/L2peaks'
```

```python
for ct in celltypes:
    # 找出包含该 celltype 前缀的所有列（例如 AMY_Hcrtr2_HapIn1_GABA-0-2）
    matching_cols = [col for col in peaks.columns if col.startswith(ct)]

    # 得到 TRUE 的 rows
    true_rows = peaks["Peaks"][peaks[matching_cols].any(axis=1)]
    true_rows.str.split(r'[-:]', expand=True).to_csv(f'/data2st1/junyi/output/atac1112/cCRE/L2peaks/{ct}_peaks.bed', sep='\t', header=False, index=False)
```
