```python
import scanpy as sc
import numpy as np
```

```python
adata = sc.read_h5ad('/data2st1/junyi/output/atac1112/3REGIONS_peak.h5ads')
```

```python
adata.obs['celltype.L4_sample'] = adata.obs['celltype.L4'].astype(str) + '_' + adata.obs['sample'].astype(str)
```

```python
adata.obs
```

```python
adata_agg = sc.get.aggregate(adata, by='celltype.L4_sample', layer='count', func=["sum", "count_nonzero"])
```

```python
adata_agg.X = adata_agg.layers['sum']
```

```python
adata_agg.X
```

```python
sc.pp.filter_cells(adata_agg, min_genes=100)
sc.pp.filter_genes(adata_agg, min_cells=3)
```

```python
sc.pp.normalize_total(adata_agg)
# Logarithmize the data
sc.pp.log1p(adata_agg)
```

```python
adata_agg.obs
```

```python
adata_agg.obs['celltype.L2'] = adata_agg.obs['celltype.L4_sample'].str[:-14]
```

```python
adata_agg.obs['sample'] = adata_agg.obs['celltype.L4_sample'].str[-9:]
```

```python
adata_agg.obs['sample_celltype.L2'] = adata_agg.obs['sample'] + '_' + adata_agg.obs['celltype.L2']
```

```python
adata_agg.write_h5ad('/data2st1/junyi/output/atac1112/3REGIONS_peak_celltypeL4_agg.h5ads')
```

```python
adata_agg = adata_agg[adata_agg.obs['celltype.L2']!='Lymphocyte', :]
```

```python
sc.tl.rank_genes_groups(adata_agg, groupby="celltype.L2", method="wilcoxon")
```

```python
adata_agg.obs
```

```python
for ct in adata_agg.obs['celltype.L2'].unique():
    df_cts = sc.get.rank_genes_groups_df(adata_agg, group=ct,pval_cutoff=0.05)
    df_cts.to_csv(f'/data2st1/junyi/output/atac1112/dar/cts/wilcoxon/{ct}.csv', index=False)
```

```python
import glob
df_files = glob.glob('/data2st1/junyi/output/atac1112/dar/cts/wilcoxon/*.csv')
```

```python
import pandas as pd
import os
import pybedtools

for file in df_files:
    df_cts = pd.read_csv(file)
    df_selected = df_cts[(df_cts['logfoldchanges']>0.1)]
    filename = os.path.basename(file).replace('.csv','.bed')
    if df_selected.shape[0]==0:
        continue
    df_bed = df_selected['names'].str.split(r'[-:]', expand=True)
    bed_object = pybedtools.BedTool.from_dataframe(df_bed)
    bed_object.sort().saveas(f'/data2st1/junyi/output/atac1112/cCRE/L2ctspeaks/{filename}')
    #sort_values(by=[0,1]).to_csv(f'/data2st1/junyi/output/atac1112/cCRE/L2ctspeaks/{filename}',sep='\t', header=False, index=False)
```

```python
target = 20000
for region in ['PFC','AMY','HIP']:
    obs_region= adata.obs[adata.obs['Region']==region]
    n = obs_region.shape[0]
    if n > target:
        idx = np.random.choice(n, target, replace=False)
        idx_obs = obs_region.index[idx]
    adata_region = adata[idx_obs,:].to_memory()
    adata_region.write_h5ad(f'/data2st1/junyi/output/atac1112/subset/{region}_20k.h5ad')
```

```python
target = 2000
for region in ['PFC','AMY','HIP']:
    adata_region = sc.read_h5ad(f'/data2st1/junyi/output/atac1112/subset/{region}_20k.h5ad',backed='r')
    obs= adata_region.obs

    n = obs.shape[0]
    if n > target:
        idx = np.random.choice(n, target, replace=False)
        idx_obs = obs.index[idx]
    adata_region_sub = adata_region[idx_obs,:].to_memory()
    adata_region_sub.write_h5ad(f'/data2st1/junyi/output/atac1112/subset/region_nt/{region}_{region}_2k.h5ad')
```
